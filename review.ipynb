```python
import pandas as pd
```

```python
data=pd.read_csv('4431229.csv')
name=data.NAME[0]
name
```

```python
station=data.STATION[0]
station
```

```python
data['DATE']=pd.to_datetime(data['DATE'])
data['TMAX']=pd.to_numeric(data['TMAX'], errors='coerce')
midway_Max_temp=data[['DATE','TMAX']].set_index('DATE')
display(midway_Max_temp.head(5))
display(midway_Max_temp.tail(5))
```

```python
midway_Max_temp.shape
```

```python
# fill in any missing calendar days (2017-08-16) as NaN
midway_Max_temp = midway_Max_temp.asfreq('D')
print(midway_Max_temp.shape)
print(midway_Max_temp[midway_Max_temp['TMAX'].isna()])

# NaN is written as "NA" in the CSV
midway_Max_temp.to_csv('midway_Max_temp.csv', na_rep='NA')
```
